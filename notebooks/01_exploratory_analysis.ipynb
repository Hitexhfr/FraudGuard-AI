```python
import pandas as pd
import numpy as np

# Load dataset
data_path = '../data/raw/PS_20174371424001719_log.csv'
df = pd.read_csv(data_path)

# 1. Dataset Dimensions
print("=== 1. DATASET OVERVIEW ===")
print(f"Total Transactions: {df.shape[0]:,}")
print(f"Total Features: {df.shape[1]}")

# 2. Null Values Check
print("\n=== 2. MISSING VALUES ===")
print(df.isnull().sum())

# 3. Target Variable Analysis
print("\n=== 3. CLASS IMBALANCE (isFraud) ===")
counts = df['isFraud'].value_counts()
pcts = df['isFraud'].value_counts(normalize=True) * 100
print(pd.DataFrame({'Count': counts, 'Percentage (%)': pcts.round(4)}))

# 4. Fraud by Transaction Type
print("\n=== 4. FRAUD BY TRANSACTION TYPE ===")
summary = df.groupby('type').agg(
    Total_Count=('isFraud', 'count'),
    Fraud_Count=('isFraud', 'sum')
)
summary['Fraud_Rate (%)'] = (summary['Fraud_Count'] / summary['Total_Count'] * 100).round(4)
print(summary)
```
